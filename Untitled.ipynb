```python
import os
import cv2
import numpy as np
import tensorflow as tf
from ultralytics import YOLO
from fast_plate_ocr.train.model.layers import MaxBlurPooling2D

yolo_model = YOLO(r"F:\models\plate_detector_best.pt")

ocr_model = tf.keras.models.load_model(
    r"F:\models\ocr_best.keras",
    compile=False,
    safe_mode=False
)

print("Both models loaded successfully")
```

```python
import os

image_folder = r"F:\models\test_images"
output_folder = r"F:\models\output"

os.makedirs(output_folder, exist_ok=True)

alphabet = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ_"

image_files = [
    f for f in os.listdir(image_folder)
    if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
]

print("Total images:", len(image_files))
```

```python
for i, filename in enumerate(image_files, 1):
    image_path = os.path.join(image_folder, filename)
    image = cv2.imread(image_path)

    if image is None:
        continue

    results = yolo_model.predict(image, conf=0.25, verbose=False)
    detected = False

    for box in results[0].boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())

        crop = image[y1:y2, x1:x2]
        if crop.size == 0:
            continue

        crop = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)
        crop = cv2.resize(crop, (128, 64))
        crop = np.expand_dims(crop.astype(np.float32), axis=0)

        prediction = ocr_model.predict(crop, verbose=0)["plate"][0]

        text = "".join(
            alphabet[idx]
            for idx in np.argmax(prediction, axis=-1)
        ).rstrip("_")

        confidence = float(np.max(prediction, axis=-1).mean())

        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(
            image, text, (x1, max(y1 - 10, 25)),
            cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2
        )

        print(f"{filename} | Plate: {text} | Confidence: {confidence:.2f}")
        detected = True

    if not detected:
        print(f"{filename} | Number plate not detected")

    cv2.imwrite(os.path.join(output_folder, filename), image)
    print(f"Progress: {i}/{len(image_files)}")

print("DONE! Results saved in:", output_folder)
```
